```python
# Create empty lists for the various attributes in insurance.csv
ages = []
sexes = []
bmis = []
num_children = []
smoker_statuses = []
regions = []
insurance_charges = []
```

```python
def load_list_data(lst, csv_file, column_name):
    """A helper function to load csv data."""
    # Open csv file.
    with open(csv_file) as csv_info:
        # Read the data from the csv file.
        csv_dict = csv.DictReader(csv_info)
        # Loop through the data in each row of the csv.
        for row in csv_dict:
            # Add the data from each row to a list.
            lst.append(row[column_name])
        
        return lst
```

I decided to build a helper function to make loading data into my lists as efficient as possible. Without this function I would have to open the **insurance.csv** and rewrite the `for` loop seven times. However, with this function I can simply call `load_list_data()` each time as shown below.

```python
# Load the data into the separate lists representing the attributes held in insurance.csv
filename = 'insurance.csv'
load_list_data(ages, filename, 'age')
load_list_data(sexes, filename, 'sex')
load_list_data(bmis, filename, 'bmi')
load_list_data(num_children, filename, 'children')
load_list_data(smoker_statuses, filename, 'smoker')
load_list_data(regions, filename, 'region')
load_list_data(insurance_charges, filename, 'charges')
```

Now that I have all the data from **insurance.csv** neatly organized into labeled lists, I am ready to start my analysis. This is where I planned out what I wanted to investigate and how to perform the analysis. There are many aspects of the data I could look into, and I have decided to analyze the following attributes:
* Average age of the patients
* Number of men vs. women counted in the dataset
* Geographical location of the patients
* Average yearly medical charges of the patients
* Creating a dictionary that contains all patient information

To perform these inspections, I have built a class called `PatientsInfo` which contains fives methods:
* `analyse_ages()`
* `analyse_sexes()`
* `unique_regions()`
* `average_charges()`
* `create_dictionary()`

```python
class PatientsInfo:
    def __init__(self, patients_ages, patients_sexes, patients_bmis, patients_num_children, 
                 patients_smoker_statuses, patients_regions, patients_charges):
        """Constructor which takes in each list representing data loaded from the insurance.csv file."""
        self.patients_ages = patients_ages
        self.patients_sexes = patients_sexes
        self.patients_bmis = patients_bmis
        self.patients_num_children = patients_num_children
        self.patients_smoker_statuses = patients_smoker_statuses
        self.patients_regions = patients_regions
        self.patients_charges = patients_charges

    def analyze_ages(self):
        """A method which calculates the average ages of the patients in the insurance.csv file."""
        # Initialise the total age to zero.
        total_age = 0
        # Iterate through all of the ages in the ages list.
        for age in self.patients_ages:
            # Calculate the sum of the total ages.
            total_age += int(age)

        return ("Average Patient Age: " + str(round(total_age/len(self.patients_ages), 2)) + " years")

    def analyze_sexes(self):
        """A method which calculates the number of male and female patients in the insurance.csv file."""
        # Initialise number of male and female patients to zero.
        female_count = 0
        male_count = 0
        # Iterate through each sex in the sexes list.
        for sex in self.patients_sexes:
            # If the patient is female increment the female count by 1.
            if sex == 'female':
                female_count += 1
            # If the patient is male increment the male count by 1.
            elif sex == 'male':
                male_count += 1
                
        print("Number of female patients: ", female_count)
        print("Number of male patients: ", male_count)

    def unique_regions(self):
        """A method to identify all of the unique regions that patients are from in the insurance.csv file."""
        # Intialise an empty list of unique regions.
        unique_regions = []
        # Iterate through each region in the regions list.
        for region in self.patients_regions:
            # If the region is not already in the unique regions list
            # then add it to the unique regions list.
            if region not in unique_regions: 
                unique_regions.append(region)

        return unique_regions

    def average_charges(self):
        """A method to find the average yearly medical charges for patients in the insurance.csv file."""
        # Initalise the total charges to zero.
        total_charges = 0
        # Iterate through all of the charges in the patients charges list and
        # add each charge to the total charge.
        for charge in self.patients_charges:
            total_charges += float(charge)

        return ("Average Yearly Medical Insurance Charges: " +  
                str(round(total_charges/len(self.patients_charges), 2)) + " dollars.")
    
    def create_dictionary(self):
        """
        A method to create a dictionary of patient data consisting of all of 
        the features loaded from the insurance.csv file.
        """
        self.patients_dictionary = {}
        self.patients_dictionary["age"] = [int(age) for age in self.patients_ages]
        self.patients_dictionary["sex"] = self.patients_sexes
        self.patients_dictionary["bmi"] = self.patients_bmis
        self.patients_dictionary["children"] = self.patients_num_children
        self.patients_dictionary["smoker"] = self.patients_smoker_statuses
        self.patients_dictionary["regions"] = self.patients_regions
        self.patients_dictionary["charges"] = self.patients_charges
        
        return self.patients_dictionary
```

I created an instance of the `patient_info` class and used each method to see the results of my analysis and interpret my findings.

```python
patient_info = PatientsInfo(ages, sexes, bmis, num_children, smoker_statuses, regions, insurance_charges)
```

```python
patient_info.analyze_ages()
```
